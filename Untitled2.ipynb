```python
import pandas as pd
import numpy as np

# 1. 重新读取数据 (确保路径正确)
df = pd.read_csv(r'C:\Users\Administrator\Desktop\homework\train_set.csv', sep='\t')

# 2. 关键步骤：将文本列(text)和标签列(label)分离出来
# X 是输入特征 (文本)
# y 是目标标签 (数字类别)
X = df['text']
y = df['label']

# 3. 验证一下 y 是否存在且正常
print("数据准备完成！")
print(f"标签 y 的前5个值: {y.head().tolist()}")
print(f"标签 y 的形状: {y.shape}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.utils.data as Data
from collections import Counter
import numpy as np
import pandas as pd
from tqdm import tqdm
import time

# ================= 1. 全局配置 =================
MAX_VOCAB_SIZE = 100000 # 词汇表最大大小
MAX_LEN = 500           # 【优化】全量数据建议缩短长度，加快训练速度
BATCH_SIZE = 64         # 批次大小
EPOCHS = 5              # 训练轮数
LEARNING_RATE = 0.001   # 学习率
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

print(f"正在使用设备: {DEVICE}")

# ================= 2. 数据加载与预处理 =================
# 假设你已经有了 df, X, y (如果没运行过，请取消下面注释)
# df = pd.read_csv(r'C:\Users\Administrator\Desktop\homework\train_set.csv', sep='\t')
# X = df['text']
# y = df['label']

# 构建词表 (Vocabulary)
def build_vocab(texts, max_size):
    counter = Counter()
    for text in texts:
        # text 是字符串 "2967 6758 339..."，需要按空格分割
        counter.update(text.split())

    # 保留最高频的词，预留 0:PAD, 1:UNK
    vocab = {'<PAD>': 0, '<UNK>': 1}
    for word, _ in counter.most_common(max_size - 2):
        vocab[word] = len(vocab)

    return vocab

print("正在构建词表...")
vocab = build_vocab(X, MAX_VOCAB_SIZE)
print(f"词表大小: {len(vocab)}")

# 文本编码函数
def encode_text(text, vocab, max_len):
    tokens = text.split()
    # 截断或填充
    if len(tokens) > max_len:
        tokens = tokens[:max_len]
    else:
        tokens = tokens + ['<PAD>'] * (max_len - len(tokens))

    # 转换为索引列表
    indices = [vocab.get(token, vocab['<UNK>']) for token in tokens]
    return indices

print("正在编码文本...")
X_encoded = [encode_text(text, vocab, MAX_LEN) for text in X]
y_labels = y.values

# ================= 3. 定义 Dataset 类 (修复 Tensor 转换问题) =================
class TextDataset(Data.Dataset):
    def __init__(self, x_data, y_data):
        self.x_data = x_data
        self.y_data = y_data

    def __len__(self):
        return len(self.y_data)

    def __getitem__(self, index):
        # 【关键修复】必须在这里将 list 转为 torch.tensor
        # 否则后续 .to(device) 会报 AttributeError
        x = torch.tensor(self.x_data[index], dtype=torch.long)
        y = torch.tensor(self.y_data[index], dtype=torch.long)
        return x, y

# 创建 DataLoader
dataset = TextDataset(X_encoded, y_labels)
train_loader = Data.DataLoader(dataset=dataset, batch_size=BATCH_SIZE, shuffle=True)

# ================= 4. 定义模型 =================
class TextClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_classes):
        super(TextClassifier, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, 128, batch_first=True, bidirectional=True)
        self.fc = nn.Linear(128 * 2, num_classes) # 双向 LSTM 维度翻倍

    def forward(self, x):
        embedded = self.embedding(x)
        # lstm_out shape: (batch, seq_len, hidden*2)
        lstm_out, (hidden, cell) = self.lstm(embedded)

        # 取最后一个时间步的输出（或者拼接 hidden state）
        # 这里简单取最后一个时刻的输出作为句子表示
        last_output = lstm_out[:, -1, :]
        logits = self.fc(last_output)
        return logits

NUM_CLASSES = 14 # 根据你的数据集标签数量修改，通常是 14
EMBED_DIM = 128

model = TextClassifier(len(vocab), EMBED_DIM, NUM_CLASSES).to(DEVICE)
print(model)

# ================= 5. 处理类别不平衡 (可选但推荐) =================
# 计算每个类别的样本数
label_counts = np.bincount(y_labels)
# 计算权重：总样本数 / (类别数 * 该类别样本数)
class_weights = 1.0 / label_counts
class_weights = class_weights / class_weights.sum() * NUM_CLASSES
class_weights = torch.FloatTensor(class_weights).to(DEVICE)

criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

# ================= 6. 训练循环 (修复 IOPub 限制) =================
print("开始全量数据训练...")
model.train()

for epoch in range(EPOCHS):
    total_loss = 0
    correct = 0
    total = 0
    start_time = time.time()

    # 【关键修复】设置 mininterval=1.0 和 ncols=80
    # 这限制了 tqdm 每秒最多刷新一次，避免触发 Jupyter 的消息速率限制
    loop = tqdm(train_loader, desc=f'Epoch {epoch+1}/{EPOCHS}', mininterval=1.0, ncols=80)

    for batch_x, batch_y in loop:
        batch_x, batch_y = batch_x.to(DEVICE), batch_y.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(batch_x)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        _, predicted = torch.max(outputs.data, 1)
        total += batch_y.size(0)
        correct += (predicted == batch_y).sum().item()

        # 更新进度条显示的 Loss 和 Acc
        loop.set_postfix(loss=loss.item(), acc=100. * correct / total)

    avg_loss = total_loss / len(train_loader)
    accuracy = 100. * correct / total
    elapsed = time.time() - start_time

    print(f"Epoch [{epoch+1}/{EPOCHS}] 耗时: {elapsed:.2f}s | Loss: {avg_loss:.4f} | Accuracy: {accuracy:.2f}%")

print("训练完成！")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.utils.data as Data
from collections import Counter
import numpy as np
import pandas as pd
from tqdm import tqdm
import time
from sklearn.model_selection import train_test_split

# ================= 1. 全局配置 =================
MAX_VOCAB_SIZE = 100000 # 词汇表最大大小
MAX_LEN = 500           # 全量数据建议缩短长度，加快训练速度
BATCH_SIZE = 64         # 批次大小
EPOCHS = 5              # 训练轮数
LEARNING_RATE = 0.001   # 学习率
NUM_CLASSES = 14        # 新闻分类类别数
EMBED_DIM = 128         # Embedding维度
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

print(f"正在使用设备: {DEVICE}")

# ================= 2. 数据加载与预处理 =================
# 假设你已经有了 df, X, y (如果没运行过，请取消下面注释)
# df = pd.read_csv(r'C:\Users\Administrator\Desktop\homework\train_set.csv', sep='\t')
# X = df['text']
# y = df['label']

# 构建词表 (Vocabulary) - 仅使用训练集构建以防信息泄露
def build_vocab(texts, max_size):
    counter = Counter()
    for text in texts:
        counter.update(text.split())
    vocab = {'<PAD>': 0, '<UNK>': 1}
    for word, _ in counter.most_common(max_size - 2):
        vocab[word] = len(vocab)
    return vocab

# 【关键修改】先按8:2划分数据，再构建词表和编码
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X, y.values, test_size=0.2, random_state=42, stratify=y.values
)
print(f"数据划分完成 | TrainSet: {len(X_train_raw)} | TestSet: {len(X_test_raw)}")

print("正在基于TrainSet构建词表...")
vocab = build_vocab(X_train_raw, MAX_VOCAB_SIZE)
print(f"词表大小: {len(vocab)}")

# 文本编码函数
def encode_text(text, vocab, max_len):
    tokens = text.split()
    if len(tokens) > max_len:
        tokens = tokens[:max_len]
    else:
        tokens = tokens + ['<PAD>'] * (max_len - len(tokens))
    indices = [vocab.get(token, vocab['<UNK>']) for token in tokens]
    return indices

print("正在编码文本...")
X_train_encoded = [encode_text(text, vocab, MAX_LEN) for text in X_train_raw]
X_test_encoded = [encode_text(text, vocab, MAX_LEN) for text in X_test_raw]

# ================= 3. 定义 Dataset 类 =================
class TextDataset(Data.Dataset):
    def __init__(self, x_data, y_data):
        self.x_data = x_data
        self.y_data = y_data

    def __len__(self):
        return len(self.y_data)

    def __getitem__(self, index):
        x = torch.tensor(self.x_data[index], dtype=torch.long)
        y = torch.tensor(self.y_data[index], dtype=torch.long)
        return x, y

# 【关键修改】分别创建训练集和测试集DataLoader
train_dataset = TextDataset(X_train_encoded, y_train)
test_dataset = TextDataset(X_test_encoded, y_test)

train_loader = Data.DataLoader(dataset=train_dataset, batch_size=BATCH_SIZE, shuffle=True)
test_loader = Data.DataLoader(dataset=test_dataset, batch_size=BATCH_SIZE, shuffle=False)

# ================= 4. 定义模型 =================
class TextClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_classes):
        super(TextClassifier, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, 128, batch_first=True, bidirectional=True)
        self.fc = nn.Linear(128 * 2, num_classes)

    def forward(self, x):
        embedded = self.embedding(x)
        lstm_out, (hidden, cell) = self.lstm(embedded)
        last_output = lstm_out[:, -1, :]
        logits = self.fc(last_output)
        return logits

model = TextClassifier(len(vocab), EMBED_DIM, NUM_CLASSES).to(DEVICE)
print(model)

# ================= 5. 处理类别不平衡 =================
# 【关键修改】仅基于TrainSet计算类别权重，防止测试集信息泄露
label_counts = np.bincount(y_train, minlength=NUM_CLASSES)
class_weights = 1.0 / label_counts
class_weights = class_weights / class_weights.sum() * NUM_CLASSES
class_weights = torch.FloatTensor(class_weights).to(DEVICE)

criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

# ================= 6. 训练循环 =================
print("开始TrainSet训练...")
model.train()

for epoch in range(EPOCHS):
    total_loss = 0
    correct = 0
    total = 0
    start_time = time.time()

    loop = tqdm(train_loader, desc=f'Epoch {epoch+1}/{EPOCHS}', mininterval=1.0, ncols=80)

    for batch_x, batch_y in loop:
        batch_x, batch_y = batch_x.to(DEVICE), batch_y.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(batch_x)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        _, predicted = torch.max(outputs.data, 1)
        total += batch_y.size(0)
        correct += (predicted == batch_y).sum().item()

        loop.set_postfix(loss=loss.item(), acc=100. * correct / total)

    avg_loss = total_loss / len(train_loader)
    accuracy = 100. * correct / total
    elapsed = time.time() - start_time

    print(f"Epoch [{epoch+1}/{EPOCHS}] 耗时: {elapsed:.2f}s | Loss: {avg_loss:.4f} | Train Acc: {accuracy:.2f}%")

print("训练完成！")

# ================= 7. 测试集评测 =================
print("\n开始在独立TestSet上评测...")
model.eval()
test_correct = 0
test_total = 0

with torch.no_grad():
    eval_loop = tqdm(test_loader, desc='Evaluating TestSet', mininterval=1.0, ncols=80)
    for batch_x, batch_y in eval_loop:
        batch_x, batch_y = batch_x.to(DEVICE), batch_y.to(DEVICE)
        outputs = model(batch_x)
        _, predicted = torch.max(outputs.data, 1)
        test_total += batch_y.size(0)
        test_correct += (predicted == batch_y).sum().item()

test_acc = 100. * test_correct / test_total
print(f"✅ TestSet Accuracy: {test_acc:.2f}%")
```
